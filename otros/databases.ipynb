```python
import pandas as pd
from pathlib import Path

# Carpeta donde están los .txt
carpeta_datos = Path(r"C:\Users\juanl\OneDrive\Developer\TFM\FinancialPhraseBank-v1.0\FinancialPhraseBank-v1.0")

# Carpeta donde quieres guardar el CSV
carpeta_salida = Path(r"C:\Users\juanl\OneDrive\Developer\TFM")

ficheros = {
    "50agree":  "Sentences_50Agree.txt",
    "66agree":  "Sentences_66Agree.txt",
    "75agree":  "Sentences_75Agree.txt",
    "allagree": "Sentences_AllAgree.txt",
}

dfs = []
for nivel, fichero in ficheros.items():
    temp = pd.read_csv(
        carpeta_datos / fichero,
        sep="@",
        header=None,
        names=["sentence", "label"],
        encoding="latin-1",
        engine="python",
    )
    temp["agreement_level"] = nivel
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)

# Guardar en la carpeta TFM
salida = carpeta_salida / "financial_phrasebank.csv"
df.to_csv(salida, index=False, encoding="utf-8")

print(f"Guardado en: {salida}")
print(f"Shape: {df.shape}")
print("\nFrases por nivel de acuerdo:")
print(df["agreement_level"].value_counts())
print("\nDistribución de etiquetas (global):")
print(df["label"].value_counts())
df.head()
```
